```python
import json

with open("train_idxs.json") as fh:
    train_idxs = {int(key): value for key, value in json.load(fh).items()}

with open("test_idxs.json") as fh:
    test_idxs = {int(key): value for key, value in json.load(fh).items()}
```

```python
from numpy.lib.format import open_memmap

scores = open_memmap("./musicgen_trak.mmap")
```

```python
scores
```

```python
from IPython.display import Audio, display
from pathlib import Path

i = 20
scores[:, 21].argsort()[-5:][::-1]
test_sample_path = Path(test_idxs[i])
test_sample_local_path = Path("data/musiccaps/music_data") / test_sample_path.name
print("TEST SAMPLE:", test_sample_path)
display(Audio(filename=test_sample_local_path, embed=True))

generated_sample_path = Path("results/generated/musicgen") / f"{i}.wav" 
print("GENERATED SAMPLE:", generated_sample_path)
display(Audio(filename=generated_sample_path, embed=True))

top_trak_scorers = scores[:, i].argsort()[-5:][::-1]
for ii, train_im_ind in enumerate(top_trak_scorers):
    train_sample_path = Path(train_idxs[train_im_ind])
    print(f"train sample {ii}: {train_sample_path}")
    train_sample_local_path = Path("data/musiccaps/music_data") / train_sample_path.name
    display(Audio(filename=train_sample_local_path, embed=True))
```
